```python
from collections import deque

grafo = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F', 'G'],
    'D': ['B', 'H'],
    'E': ['B', 'H', 'I'],
    'F': ['C', 'I', 'J'],
    'G': ['C', 'J'],
    'H': ['D', 'E', 'I'],
    'I': ['E', 'F', 'H', 'J'],
    'J': ['F', 'G', 'I']
}

def bfs(grafo, inicio):
    visitados = [inicio]
    cola = deque([inicio])
    secuencia = []
    paso = 1

    print("=== BFS ===")
    while cola:
        actual = cola.popleft()
        secuencia.append(actual)
        for vecino in sorted(grafo[actual]):
            if vecino not in visitados:
                visitados.append(vecino)
                cola.append(vecino)
        print(f"Etapa {paso:2d}  Actual: {actual}  Cola: {list(cola)}  Visitados: {visitados}")
        paso += 1
    return secuencia

def dfs(grafo, inicio):
    visitados = []
    pila_llamadas = []
    paso = [1]

    def explorar(u):
        visitados.append(u)
        pila_llamadas.append(u)
        print(f"Etapa {paso[0]:2d}  Actual: {u}  Pila: {pila_llamadas}  Visitados: {visitados}")
        paso[0] += 1
        for v in sorted(grafo[u]):
            if v not in visitados:
                explorar(v)
        pila_llamadas.pop()

    print("\n=== DFS ===")
    explorar(inicio)
    print(f"Etapa {paso[0]:2d}  Retorno completo  Pila: {pila_llamadas}")
    return visitados

if __name__ == "__main__":
    seq_bfs = bfs(grafo, 'A')
    print("Secuencia Final BFS:", " -> ".join(seq_bfs))

    seq_dfs = dfs(grafo, 'A')
    print("Secuencia Final DFS:", " -> ".join(seq_dfs))
```
