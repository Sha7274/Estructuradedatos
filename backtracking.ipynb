```python
def cargar_laberinto(nombre_archivo):
    with open(nombre_archivo, 'r') as archivo:
        filas = int(archivo.readline().strip())
        columnas = int(archivo.readline().strip())
        laberinto = []
        for linea in archivo:
            valores = linea.strip().split(',')
            if len(valores) > 1:
                laberinto.append(valores)
    return laberinto, filas, columnas

def encontrar_inicio(laberinto):
    for f in range(len(laberinto)):
        for c in range(len(laberinto[0])):
            if laberinto[f][c] == 'S':
                return f, c
    return None

def resolver_laberinto(laberinto):
    inicio = encontrar_inicio(laberinto)
    if not inicio:
        return "No se encontró el inicio"

    pila = [inicio]
    visitados = set()
    visitados.add(inicio)
    movimientos = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    while pila:
        f, c = pila[-1]

        if laberinto[f][c] == 'E':
            return pila

        avanzamos = False

        for df, dc in movimientos:
            nf, nc = f + df, c + dc
            if 0 <= nf < len(laberinto) and 0 <= nc < len(laberinto[0]):
                if laberinto[nf][nc] in ['0', 'E'] and (nf, nc) not in visitados:
                    pila.append((nf, nc))
                    visitados.add((nf, nc))
                    avanzamos = True
                    break

        if not avanzamos:
            pila.pop()

    return None

def imprimir_laberinto_resuelto(laberinto, camino):
    for f, c in camino:
        if laberinto[f][c] not in ['S', 'E']:
            laberinto[f][c] = '-'

    for fila in laberinto:
        print(" ".join(fila))

if __name__ == "__main__":
    archivo_prueba = 'entrada.lab'
    laberinto_2d, f, c = cargar_laberinto(archivo_prueba)

    camino = resolver_laberinto(laberinto_2d)

    if camino:
        print(f"camino encontrado {archivo_prueba}:")
        imprimir_laberinto_resuelto(laberinto_2d, camino)
    else:
        print(f"No se encontró solución para {archivo_prueba}.")
```
