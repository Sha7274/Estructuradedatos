```python
class WebAnalytics:
    def __init__(self):
        self._unique_visitors = set()

    def register_visit(self, ip_address: str):
        self._unique_visitors.add(ip_address)

    def get_unique_visitor_count(self) -> int:
        return len(self._unique_visitors)

    def get_unique_visitors(self) -> set:
        return self._unique_visitors


if __name__ == "__main__":
    print("=== PRUEBA DE CASO DE USO: Visitantes Únicos ===")

    log_visitas = [
        "192.168.1.10",
        "192.168.1.25",
        "192.168.1.10",
        "192.168.1.32",
        "192.168.1.25"
    ]

    analytics = WebAnalytics()

    print("\n1. Registrando accesos al servidor web:")
    for ip in log_visitas:
        analytics.register_visit(ip)
        print(f"   -> Acceso detectado desde la IP: {ip}")

    print("\n2. Resultados de las pruebas:")
    print(f"   - Total de registros procesados: {len(log_visitas)}")
    print(f"   - Total de visitantes únicos calculados: {analytics.get_unique_visitor_count()}")
    print(f"   - Direcciones IP únicas almacenadas: {list(analytics.get_unique_visitors())}")

    esperados = 3
    actual = analytics.get_unique_visitor_count()
    if actual == esperados:
        print("\n[ÉXITO] La prueba pasó correctamente: se identificaron 3 visitantes únicos.")
    else:
        print(f"\n[FALLÓ] Se esperaban {esperados} visitantes, pero se obtuvieron {actual}.")
```
