```python
def dos_suma(nums, objetivo):
    """
    Encuentra los índices de dos números que suman el valor objetivo
    utilizando una Tabla Hash (diccionario en Python).
    """
    tabla_hash = {}

    for indice, numero in enumerate(nums):
        complemento = objetivo - numero

        if complemento in tabla_hash:
            return [tabla_hash[complemento], indice]

        tabla_hash[numero] = indice

    return []

if __name__ == "__main__":
    arreglo = [2, 14, 6, 11, 8, 15, 7]
    objetivo = 9

    resultado = dos_suma(arreglo, objetivo)

    print("=" * 45)
    print("        RESULTADO DEL PROBLEMA HASH         ")
    print("=" * 45)
    print(f"Arreglo de entrada: {arreglo}")
    print(f"Objetivo de suma:   {objetivo}")
    print("-" * 45)

    if resultado:
        idx1, idx2 = resultado
        val1, val2 = arreglo[idx1], arreglo[idx2]
        print(f"Índices encontrados: {resultado}")
        print(f"Comprobación:        arreglo[{idx1}] + arreglo[{idx2}] -> {val1} + {val2} = {val1 + val2}")
    else:
        print("No se encontró ningún par que sume el objetivo.")
    print("=" * 45)
```
